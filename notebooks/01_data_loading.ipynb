```python
sales["Transaction Date"] = pd.to_datetime(
    sales["Transaction Date"], errors="coerce"
)
```

```python
print("Sales start date:", sales["Transaction Date"].min())
print("Sales end date:", sales["Transaction Date"].max())
print(
    "Invalid sales dates:",
    sales["Transaction Date"].isna().sum()
)
```

```python
# 2. Inventory date range
inventory["Start Date"] = pd.to_datetime(
    inventory["Start Date"],
    errors="coerce"
)

inventory["End Date"] = pd.to_datetime(
    inventory["End Date"],
    errors="coerce"
)
```

```python
print("\nInventory start date:", inventory["Start Date"].min())
print("Inventory end date:", inventory["End Date"].max())

print(
    "Invalid inventory start dates:",
    inventory["Start Date"].isna().sum()
)

print(
    "Invalid inventory end dates:",
    inventory["End Date"].isna().sum()
)
```

```python
# 3. Stock status
print("\nStock status counts:")
display(inventory["Stock Status"].value_counts(dropna=False))
```

```python
# 4. Negative stock by status
print("\nNegative stock by status:")
display(
    inventory[inventory["Qty on hand"] < 0]
    .groupby("Stock Status")
    .agg(
        records=("Qty on hand", "size"),
        total_quantity=("Qty on hand", "sum")
    )
    .sort_values("records", ascending=False)
)
```

```python
negative_stock = inventory[
    inventory["Qty on hand"] < 0
]
```

```python
# Most affected products
print("Top 10 products with negative stock:")

display(
    negative_stock.groupby(
        ["Product No", "Product Description"]
    )
    .agg(
        negative_records=("Qty on hand", "size"),
        total_stock=("Qty on hand", "sum")
    )
    .sort_values("negative_records", ascending=False)
    .head(10)
)
```

```python
# Most affected stores
print("Top 10 stores with negative stock:")

display(
    negative_stock.groupby("Store")
    .agg(
        negative_records=("Qty on hand", "size"),
        total_stock=("Qty on hand", "sum")
    )
    .sort_values("negative_records", ascending=False)
    .head(10)
)
```

```python
# Negative stock value check
print("Negative stock quantity:", (inventory["Qty on hand"] < 0).sum())

print(
    "Negative cost of stocks:",
    (inventory["Cost of Stocks"] < 0).sum()
)

print(
    "Negative stock selling amount:",
    (inventory["Stocks Selling Amount"] < 0).sum()
)
```

```python
# Return transaction summary
return_summary = (
    sales.groupby("Is Return")
    .agg(
        records=("Is Return", "size"),
        quantity=("Qty Sold", "sum"),
        sales_amount=("Sales Amount", "sum"),
        cogs=("Cogs", "sum")
    )
)

display(return_summary)
```
