## Step 10 — Overall Sales Performance Analysis

### Step 10.1: Overall Sales Performance

Calculate key sales metrics to understand the overall performance of the retail business, including sales value, quantity sold, return activity, cost of goods sold, and estimated gross profit.

```python
total_records  = len(sales)
```

```python
# Separate normal sales and return-marked records
normal_sales = sales[sales["Is Return"] == 0]
returns = sales[sales["Is Return"] == 1]
```

```python
# Calculate overall metrics
total_sales_records = len(normal_sales)
total_return_records = len(returns)

total_sales_amount = normal_sales["Sales Amount"].sum()
total_return_amount = returns["Sales Amount"].sum()

total_qty_sold = normal_sales["Qty Sold"].sum()
total_return_qty = returns["Qty Sold"].sum()

total_cogs = normal_sales["Cogs"].sum()
gross_profit = total_sales_amount - total_cogs

return_recod_percentage = ( total_return_records / total_records) * 100
```

```python
print("Total sales dataset records:", total_records)
print("Normal sales records:", total_sales_records)
print("Return-marked records:", total_return_records)

print(f"\nNormal sales amount: ₹{total_sales_amount:,.2f}")
print(f"Return amount: ₹{total_return_amount:,.2f}")

print(f"\nQuantity sold: {total_qty_sold:,.0f}")
print(f"Returned quantity: {total_return_qty:,.0f}")

print(f"\nNormal sales COGS: ₹{total_cogs:,.2f}")
print(f"Estimated gross profit: ₹{gross_profit:,.2f}")

print(
    f"\nReturn-marked records: "
    f"{return_recod_percentage:.2f}% of all records"
)
```

## Step 10.2 Monthly Sales Trend
# "Month-wise sales value kaise change hui, aur Kaunse month mein sales sabse zyada thi?"
# Analyze monthly sales value and quantity sold to identify high-performing and low-performing months. Return-marked records are excluded from this initial normal-sales trend and will be analyzed separately.

```python
# keep normal sales records
monthly_sales_data = sales[sales["Is Return"] == 0].copy()
```

```python
# Extract month from transaction date
monthly_sales_data["Month"]= (
    monthly_sales_data["Transaction Date"].dt.to_period("M")
)
```

```python
# Aggregate monthly metrics
monthly_sales = (
    monthly_sales_data
    .groupby("Month")
    .agg(
        Sales_Records=("Sales Amount", "size"),
        Total_Revenue=("Sales Amount", "sum"),
        Total_Quantity=("Qty Sold", "sum"),
        Total_COGS=("Cogs", "sum")
    )
    .reset_index()
)
```

```python
# Calculate estimated gross profit
monthly_sales["Estimated_Gross_Profit"] = (
    monthly_sales["Total_Revenue"]
    - monthly_sales["Total_COGS"]
)
```

```python
# Display results
display(monthly_sales.round(2))
```
